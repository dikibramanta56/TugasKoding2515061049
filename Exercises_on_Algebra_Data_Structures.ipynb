what is the transpose of this vector?
[25
2
-3
-23
]
JAWABAN:
[25 2 -3 -23]


```python
import numpy as np

# Membuat vektor kolom
vektor = np.array([[25],
                   [2],
                   [-3],
                   [-23]])

# Transpose
transpose_vektor = vektor.T

# Menampilkan hasil
print(transpose_vektor)
```
